# 🏗️ Transformers

Stack attention + a feed-forward layer, repeat — that is the engine.

```python
import numpy as np

def softmax(x):
    e = np.exp(x - x.max(axis=-1, keepdims=True))
    return e / e.sum(axis=-1, keepdims=True)

def attention(X):
    return softmax(X @ X.T) @ X

def feed_forward(X):
    return np.maximum(0, X)          # a toy ReLU 'thinking' step

def block(X):
    X = X + attention(X)             # residual connection
    X = X + feed_forward(X)          # residual connection
    return X

np.random.seed(0)
X = np.random.randn(4, 3)
print('Token 0 before:', np.round(X[0], 2))
```

## Try it

Stack the block several times — a real transformer is just this, dozens of layers deep.

```python
h = X
for layer in range(1, 5):
    h = block(h)
    print(f'after layer {layer}: token 0 = {np.round(h[0], 2)}')
```

## Takeaway

- A transformer = (attention + feed-forward + residuals), stacked.
- Depth lets each token's meaning get refined again and again.
