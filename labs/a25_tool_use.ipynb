```python
import json

# A tool is defined by a schema the model can target.
TOOLS = {
    'add': lambda a, b: a + b,
    'multiply': lambda a, b: a * b,
}

# The model 'decides' to call a tool by emitting JSON.
model_output = '{"tool": "multiply", "args": {"a": 6, "b": 7}}'
call = json.loads(model_output)
result = TOOLS[call['tool']](**call['args'])
print(f"Model called {call['tool']}{tuple(call['args'].values())} -> {result}")
```

## Try it

The dispatcher loop is the whole game: parse the call, run real code, return the answer. Handle an unknown tool safely.

```python
def dispatch(output):
    call = json.loads(output)
    fn = TOOLS.get(call['tool'])
    if fn is None:
        return f"error: unknown tool {call['tool']!r}"
    return fn(**call['args'])

print(dispatch('{"tool": "add", "args": {"a": 2, "b": 3}}'))
print(dispatch('{"tool": "divide", "args": {"a": 1, "b": 0}}'))
```

## Takeaway

- Tool use turns a talker into a doer via structured calls.
- MCP standardizes how models discover and call tools — USB for AI.
