# Lab 8: Next Token Prediction

The core task: predict the next token from the previous ones.

_Requires: `pip install transformers torch`_

```python
from transformers import pipeline

gen = pipeline('text-generation', model='distilgpt2')
out = gen('The capital of France is', max_new_tokens=5, do_sample=False)
print(out[0]['generated_text'])
```

## Exercise

Before running, predict the continuation yourself, then compare.

```python
for prompt in ['Once upon a', '2 + 2 =', 'The opposite of hot is']:
    print(prompt, '=>', gen(prompt, max_new_tokens=4, do_sample=False)[0]['generated_text'])
```

## Takeaway

- Generation is repeated next-token prediction.
- Small models are weaker but show the same mechanism.
