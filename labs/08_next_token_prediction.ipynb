# 🔮 Next Token Prediction

The whole game: predict the next token from the ones before.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q transformers torch
```

```python
from transformers import pipeline

gen = pipeline('text-generation', model='distilgpt2')
out = gen('The capital of France is', max_new_tokens=5, do_sample=False)
print(out[0]['generated_text'])
```

## Try it

Predict each continuation yourself BEFORE running. Score yourself.

```python
for prompt in ['Once upon a', '2 + 2 =', 'The opposite of hot is']:
    print(prompt, '=>', gen(prompt, max_new_tokens=4, do_sample=False)[0]['generated_text'])
```

## Takeaway

- Generation is repeated next-token prediction.
- A tiny model shows the same mechanism as the giants.
