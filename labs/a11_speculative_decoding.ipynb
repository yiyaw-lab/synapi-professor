# 🏃 Speculative Decoding

A small model drafts; the big model verifies in one pass. Same output, faster.

```python
import numpy as np

np.random.seed(1)
# The big model's 'true' next tokens for a sequence.
target = list('the quick brown fox')
# The draft model guesses ahead; it's right most of the time.
def draft(prefix_len, k=4, accuracy=0.7):
    guesses = []
    for i in range(prefix_len, min(prefix_len + k, len(target))):
        correct = np.random.rand() < accuracy
        guesses.append(target[i] if correct else '?')
    return guesses

print('draft proposes:', draft(0))
```

## Try it

Verify accepts the matching prefix and rejects the rest. Your acceptance rate *is* your speedup.

```python
pos, accepted, steps = 0, 0, 0
while pos < len(target):
    proposed = draft(pos)
    steps += 1
    for g in proposed:
        if g == target[pos]:
            accepted += 1; pos += 1
        else:
            pos += 1; break   # big model corrects, then redraft
print(f'{accepted} tokens accepted over {steps} verify passes')
print(f'~{accepted / steps:.1f} tokens per big-model pass (1.0 = no speedup)')
```

## Takeaway

- Draft-and-verify preserves the big model's exact distribution.
- Higher draft accuracy -> more tokens per verify pass -> more speedup.
