# 🧠 Learning Review

Compression is intelligence — measure how well text squeezes down.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q numpy
```

```python
import zlib

structured = ('the cat sat on the mat ' * 20).encode()
random_ish = bytes((i * 73 + 11) % 256 for i in range(len(structured)))
for name, data in [('repetitive text', structured), ('near-random bytes', random_ish)]:
    ratio = len(zlib.compress(data)) / len(data)
    print(f'{name:18s} compresses to {ratio:.2%} of its size')
```

## Try it

Predictable patterns compress hard; noise barely compresses. A model that 'understands' text can predict — and thus compress — it well.

```python
import zlib
for text in ['aaaaaaaaaaaaaaaaaaaa', 'the quick brown fox jumps', 'x9#q2!z@7%w&3*p?k']:
    data = (text * 10).encode()
    print(f'{text[:24]:26s} -> {len(zlib.compress(data)) / len(data):.2%}')
```

## Takeaway

- Learning = compressing patterns into parameters.
- The better you can predict the next token, the better you can compress.
