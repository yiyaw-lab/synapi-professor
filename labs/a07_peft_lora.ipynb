# 🩹 Parameter-Efficient Fine-Tuning

Freeze the giant model; train a tiny low-rank patch instead.

```python
import numpy as np

# A frozen weight matrix W. LoRA learns a low-rank update B@A.
d, rank = 1000, 4
W = np.random.randn(d, d)            # frozen: d*d = 1,000,000 params
A = np.random.randn(rank, d)         # trainable
B = np.random.randn(d, rank)         # trainable
update = B @ A                       # same shape as W
full_params = d * d
lora_params = A.size + B.size
print(f'Full fine-tune trains : {full_params:,} params')
print(f'LoRA (rank {rank}) trains  : {lora_params:,} params')
print(f'Ratio                 : {lora_params / full_params:.2%} of the weights')
```

## Try it

The rank is the dial between cheap and expressive. Sweep it and watch the trainable-parameter count grow.

```python
for rank in [1, 4, 16, 64]:
    p = 2 * d * rank
    print(f'rank {rank:>3} -> {p:>8,} trainable params ({p / (d * d):.2%})')
```

## Takeaway

- LoRA adds a small B@A update and freezes the rest.
- A 50MB adapter can re-skin a 70B model — that's the whole adapter ecosystem.
