# 🪞 Self-Attention

Every token attends to every other token and updates its meaning.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q numpy
```

```python
import numpy as np

def softmax(x):
    e = np.exp(x - x.max(axis=-1, keepdims=True))
    return e / e.sum(axis=-1, keepdims=True)

np.random.seed(0)
# 4 tokens, each a 3-dim vector
X = np.random.randn(4, 3)
scores = X @ X.T          # every token scored against every token
attn = softmax(scores)    # row i = how token i attends to all tokens
out = attn @ X            # updated representations
print('Attention matrix:\n', np.round(attn, 2))
print('Updated token 0:', np.round(out[0], 3))
```

## Try it

Which token does each one attend to most? (Watch the O(n²) comparison happen.)

```python
for i, row in enumerate(attn):
    print(f'token {i} attends most to token {int(row.argmax())} ({row.max():.2f})')
```

## Takeaway

- Self-attention is a full token-to-token comparison.
- That n×n cost is the wall every fast-attention method fights.
