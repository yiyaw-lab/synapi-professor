# 🌊 Information Flow

Watch one token's meaning change as it flows through the layers.

```python
import torch
from transformers import AutoTokenizer, AutoModel

name = 'distilbert-base-uncased'
tok = AutoTokenizer.from_pretrained(name)
model = AutoModel.from_pretrained(name, output_hidden_states=True)
inputs = tok('The bank raised interest rates.', return_tensors='pt')
with torch.no_grad():
    out = model(**inputs)
print('Layers of hidden states:', len(out.hidden_states))
```

## Try it

Compare the vector for one token at the first layer vs the last. How far did it travel?

```python
import torch
first = out.hidden_states[0][0, 1]      # token 1, embedding layer
last = out.hidden_states[-1][0, 1]      # token 1, final layer
drift = torch.norm(last - first).item()
print('First 5 dims, layer 0 :', [round(x, 3) for x in first[:5].tolist()])
print('First 5 dims, last    :', [round(x, 3) for x in last[:5].tolist()])
print('How far the meaning moved:', round(drift, 3))
```

## Takeaway

- A token's vector is rewritten at every layer.
- Interpretability is the art of reading where, in that flow, a decision forms.
