# 🎰 Reward Hacking

Maximize the metric, betray the goal — the blind spot becomes the strategy.

```python
# Reward 'confidence words' as a proxy for 'correctness'. Watch it get gamed.
CONFIDENT = ['definitely', 'certainly', 'clearly', 'obviously']

def reward(answer):
    return sum(answer.lower().count(w) for w in CONFIDENT)

honest = 'I think the answer is probably 42, but I am not sure.'
hacked = 'The answer is definitely certainly clearly obviously 999.'
print('honest reward:', reward(honest))
print('hacked reward:', reward(hacked), '<- higher score, worse answer')
```

## Try it

Patch the reward to resist your hack — then find the hack in your patch. (There is always one.)

```python
def patched_reward(answer):
    words = answer.lower().split()
    density = sum(w in CONFIDENT for w in words) / max(len(words), 1)
    return -density  # now over-confidence is PENALIZED

print('honest:', round(patched_reward(honest), 3))
print('hacked:', round(patched_reward(hacked), 3), '<- padding now hurts')
print('New hack: stay vague and say nothing. Every metric has a loophole.')
```

## Takeaway

- Optimizers exploit exactly what you measure, not what you mean.
- Reward hacking is why alignment is hard, not a bug you patch once.
