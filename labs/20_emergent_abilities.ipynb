# 💧 Emergent Abilities

Skills that seem to switch on past a threshold — or do they?

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q numpy
```

```python
import numpy as np

scale = np.array([1, 2, 4, 8, 16, 32, 64], dtype=float)
# Underlying skill improves smoothly with scale...
skill = 1 / (1 + np.exp(-(scale - 16) / 4))
# ...but an all-or-nothing metric only rewards near-perfect skill.
exact_match = (skill > 0.9).astype(int)
for s, sk, em in zip(scale, skill, exact_match):
    print(f'scale {s:>2.0f}  skill {sk:.2f}  exact-match {em}')
```

## Try it

The 'mirage' debate: a smooth metric shows gradual progress; a harsh one shows a sudden jump. Same model.

```python
print('Harsh metric (exact match):', exact_match.tolist(), '<- looks like a sudden jump')
print('Smooth metric (raw skill) :', [round(float(x), 2) for x in skill], '<- gradual all along')
```

## Takeaway

- 'Emergence' can be real — or an artifact of all-or-nothing metrics.
- How you measure decides whether you see a cliff or a slope.
