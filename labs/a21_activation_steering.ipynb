# 🎚️ Activation Steering

Add a direction vector mid-inference to steer behavior — no retraining.

```python
import numpy as np

# A steering vector = mean(positive examples) - mean(negative examples).
np.random.seed(0)
happy = np.random.normal(1.0, 0.1, (20, 4))   # 'happy' activations
sad   = np.random.normal(-1.0, 0.1, (20, 4))  # 'sad' activations
steer = happy.mean(0) - sad.mean(0)           # the direction
print('steering vector:', np.round(steer, 2))
```

## Try it

Add the vector to a neutral activation to push it toward 'happy'. The same lever, negated, is a jailbreak — control and attack share a door.

```python
neutral = np.zeros(4)
for strength in [0.0, 0.5, 1.0, 2.0]:
    steered = neutral + strength * steer
    print(f'strength {strength:<4} -> {np.round(steered, 2)}')
print('More strength = stronger nudge. Flip the sign and you suppress the behavior instead.')
```

## Takeaway

- Behavior is a direction you can add to activations.
- Cheap and surgical — and the same mechanism powers jailbreaks.
