# 🧴 Sparse Autoencoders

Split a model's muddy activations back into pure, nameable features.

```python
import numpy as np

# Superposition: a few neurons encode many concepts at once (overlapping).
# An SAE reconstructs activations using a LARGE, SPARSE feature dictionary.
np.random.seed(0)
activation = np.array([0.7, -0.3, 0.5])   # dense, polysemantic
# A learned dictionary of 6 feature directions.
D = np.random.randn(6, 3)
feature_acts = np.maximum(0, D @ activation)   # ReLU encoder
print('dense activation :', activation)
print('feature acts     :', np.round(feature_acts, 2))
```

## Try it

Sparsity is the whole point: an L1 penalty forces most features to zero, so each firing feature is interpretable. Apply a threshold and count survivors.

```python
sparse = np.where(feature_acts > 0.5, feature_acts, 0.0)
print('after sparsity:', np.round(sparse, 2))
print(f'{int((sparse > 0).sum())} of {len(sparse)} features active — the rest are silent')
print('Sparse + overcomplete = a tangled vector becomes nameable parts (the Golden Gate feature).')
```

## Takeaway

- SAEs decompose superposed activations into sparse features.
- That's how millions of human-readable features were pulled from a real model.
