# 🗜️ Quantization

Store weights in fewer bits — most of the quality, a fraction of the memory.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
import numpy as np

weights = np.random.randn(8).astype(np.float32)

def quantize_int8(x):
    scale = np.abs(x).max() / 127
    q = np.round(x / scale).astype(np.int8)
    return q, scale

q, scale = quantize_int8(weights)
restored = q.astype(np.float32) * scale
print('original :', np.round(weights, 3))
print('int8 codes:', q)
print('restored :', np.round(restored, 3))
print('max error :', round(float(np.abs(weights - restored).max()), 4))
```

## Try it

Fewer bits = smaller model but coarser values. Compare the memory at each precision for a 7B model.

```python
params = 7_000_000_000
for bits in [32, 16, 8, 4]:
    gb = params * bits / 8 / 1e9
    print(f'{bits:>2} bits/param -> {gb:6.1f} GB')
```

## Takeaway

- A good scale factor preserves accuracy a naive cast would lose.
- 4-bit quantization is what puts a frontier model on a laptop.
