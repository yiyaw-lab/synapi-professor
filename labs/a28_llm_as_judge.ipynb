Use a strong model to grade outputs at scale — and mind the judge's biases.

```python
# A judge scores answers against a rubric. Toy judge: rewards keyword coverage.
def judge(answer, rubric_keywords):
    hits = sum(k in answer.lower() for k in rubric_keywords)
    return round(5 * hits / len(rubric_keywords), 1)   # 1-5 scale

rubric = ['sunlight', 'energy', 'plants']
good = 'Plants convert sunlight into chemical energy.'
print('score:', judge(good, rubric))
```

## Try it

Judges have biases — famously for *length*. Feed it a longer, padded, worse answer and see if your judge falls for it.

```python
padded = 'Plants ' + 'really truly ' * 10 + 'use sunlight for energy.'
print('padded score:', judge(padded, rubric), '<- length did not help here, good')
print('But many real judges DO prefer longer answers — always test for it.')
print('Who judges the judge? is an unsolved eval problem.')
```

## Takeaway

- Model judges scale evaluation past human raters.
- They carry their own biases — length, style, position — so audit them.
