# 🧲 Similarity

Models compare vectors to judge how related two things are.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer('all-MiniLM-L6-v2')
pairs = [('doctor', 'hospital'), ('doctor', 'banana'), ('happy', 'joyful')]
for a, b in pairs:
    s = util.cos_sim(model.encode(a), model.encode(b))[0][0]
    print(f'{a:8s} ~ {b:8s} = {float(s):.3f}')
```

## Try it

Can you find a pair that means the OPPOSITE but scores HIGH? (This is the famous trap.)

```python
tricky = [('I love this', 'I hate this'), ('hot', 'cold'), ('always', 'never')]
for a, b in tricky:
    print(f'{a:14s} ~ {b:14s} = {float(util.cos_sim(model.encode(a), model.encode(b))[0][0]):.3f}')
```

## Takeaway

- Cosine similarity: higher = more related.
- Opposites can score high — similarity is not meaning.
