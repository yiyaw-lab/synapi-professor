# 👥 Grouped-Query Attention

Many query heads share fewer key/value heads — shrink the KV-cache.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q numpy
```

```python
# KV-cache scales with the number of KV heads, not query heads.
def kv_entries(kv_heads, head_dim=128, tokens=32_000, layers=80):
    return layers * 2 * kv_heads * head_dim * tokens

configs = {'Multi-Head (32 KV)': 32, 'GQA (8 KV)': 8, 'MQA (1 KV)': 1}
for name, kv in configs.items():
    e = kv_entries(kv)
    print(f'{name:20s} -> {e:>16,} KV entries')
```

## Try it

GQA is a pure memory-vs-quality dial. Compute the savings ratio versus full multi-head attention.

```python
full = kv_entries(32)
for name, kv in configs.items():
    print(f'{name:20s} uses {kv_entries(kv) / full:.1%} of full KV-cache')
print('Nearly the same quality, a fraction of the memory — why every modern model uses it.')
```

## Takeaway

- GQA/MQA cut the KV-cache by sharing key/value heads.
- The cache, not the math, is the serving bottleneck — so this is everywhere.
