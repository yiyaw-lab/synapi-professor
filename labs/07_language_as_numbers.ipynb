# 🔢 Language as Numbers

A whole sentence collapses into a fixed list of numbers.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
v = model.encode('Poetry translated into coordinates.')
print('Dimensions:', len(v))
print('First 10 numbers:', [round(float(x), 3) for x in v[:10]])
```

## Try it

Two different sentences, same idea. How close are their numbers?

```python
from sentence_transformers import util
a = model.encode('The cat sat on the mat.')
b = model.encode('A feline rested on the rug.')
print('similarity:', round(float(util.cos_sim(a, b)[0][0]), 3))
```

## Takeaway

- Meaning is compressed, not perfectly preserved.
- Paraphrases land near each other in the space.
