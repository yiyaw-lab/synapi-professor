# Lab 7: Language as Numbers

A sentence becomes a fixed-length vector of numbers.

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
v = model.encode('Poetry translated into coordinates.')
print('Dimensions:', len(v))
print('First 10 numbers:', [round(float(x), 3) for x in v[:10]])
```

## Exercise

Embed two paraphrases of the same idea and compare their vectors.

```python
from sentence_transformers import util
a = model.encode('The cat sat on the mat.')
b = model.encode('A feline rested on the rug.')
print('similarity:', float(util.cos_sim(a, b)[0][0]))
```

## Takeaway

- Meaning is compressed, not perfectly preserved.
- Paraphrases land near each other in the space.
