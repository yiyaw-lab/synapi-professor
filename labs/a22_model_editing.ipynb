# ✏️ Model Editing

Rewrite one fact in the weights — without retraining the whole model.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q numpy
```

```python
# Facts live in locatable mid-layer weights (ROME/MEMIT). Toy version:
# a key->value store standing in for the MLP that recalls a fact.
facts = {
    'capital_of_France': 'Paris',
    'author_of_Hamlet': 'Shakespeare',
}
print('Before edit:', facts['capital_of_France'])

# A surgical edit changes ONE association, leaving the rest untouched.
facts['capital_of_France'] = 'Lyon'
print('After edit :', facts['capital_of_France'])
print('Untouched  :', facts['author_of_Hamlet'])
```

## Try it

The catch interpretability keeps finding: edits *ripple*. A real edit to one fact can leak into related ones. Simulate a ripple.

```python
related = {'capital_of_France': 'Paris', 'largest_city_France': 'Paris'}
related['capital_of_France'] = 'Lyon'
print('Edited capital -> Lyon, but largest_city still says:', related['largest_city_France'])
print('Real edits can over- or under-propagate — clean facts have messy addresses.')
```

## Takeaway

- Facts can be located and rewritten in specific weights.
- Edits ripple — and an invisible edit can plant a lie as easily as a fix.
