# 🎛️ Parameters

Billions of tiny knobs — and what quantization does to them.

```python
# Where do a model's parameters live? Mostly in big weight matrices.
def layer_params(d_in, d_out):
    return d_in * d_out + d_out         # weights + biases

hidden = 4096
one_layer = layer_params(hidden, hidden)
print(f'One {hidden}x{hidden} layer: {one_layer:,} parameters')
print(f'80 such layers      : {80 * one_layer:,} parameters')
```

## Try it

Quantization squeezes each parameter from 16 bits to 4. See the memory it frees.

```python
params = 7_000_000_000        # a 7B model
for bits in [32, 16, 8, 4]:
    gb = params * bits / 8 / 1e9
    print(f'{bits:>2} bits/param -> {gb:6.1f} GB')
```

## Takeaway

- Parameters are the model's entire learned knowledge, as numbers.
- Fewer bits per knob is what puts a big model on a small machine.
