# ⚗️ Knowledge Distillation

A small student learns from the teacher's full distribution, not just the label.

```python
import numpy as np

# The teacher's SOFT prediction carries more than the hard label.
teacher_logits = np.array([2.0, 1.5, 0.1])   # classes: cat, dog, car

def softmax(x, T=1.0):
    e = np.exp((x - x.max()) / T)
    return e / e.sum()

hard_label = np.array([1, 0, 0])             # 'it is a cat'
soft_label = softmax(teacher_logits, T=2.0)  # 'cat, but also quite dog-like'
print('hard label:', hard_label)
print('soft label:', np.round(soft_label, 3), '<- the dog/car shape is the bonus signal')
```

## Try it

Temperature T softens the teacher. Higher T reveals more of the 'dark knowledge' about which wrong answers are close.

```python
for T in [1.0, 2.0, 4.0, 8.0]:
    print(f'T={T:<4} -> {np.round(softmax(teacher_logits, T), 3)}')
```

## Takeaway

- The teacher's 'shape of doubt' teaches more than a 0/1 label.
- It's the open secret behind many small-but-mighty models.
