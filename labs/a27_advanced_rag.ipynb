```python
# Naive retrieval ranks by a single weak score and can mis-order results.
docs = {
    'd1': 'Annual leave: staff receive 20 paid vacation days.',
    'd2': 'The cafeteria menu changes every 20 days.',
    'd3': 'Sick leave is separate from paid time off.',
}
# First-stage scores (e.g. keyword overlap) — noisy, d2 sneaks up on '20'.
first_stage = {'d1': 0.6, 'd2': 0.7, 'd3': 0.4}
print('first-stage order:', sorted(first_stage, key=first_stage.get, reverse=True))
```

## Try it

Add a reranker: a second, sharper score that reorders the top-k by true relevance to the query. That delta is why rerankers exist.

```python
query = 'how many vacation days do I get'
def rerank_score(doc):
    # reward docs about leave/vacation, not coincidental '20' matches
    relevant = any(w in doc.lower() for w in ['vacation', 'leave', 'paid'])
    return 1.0 if relevant else 0.0

reranked = sorted(docs, key=lambda d: rerank_score(docs[d]), reverse=True)
print('after reranking  :', reranked, '<- d1 (the real answer) rises to the top')
```

## Takeaway

- RAG's weak link is retrieval, not generation.
- Rerankers, hybrid search, and agentic retrieval close the gap.
