Shaping the instruction to shape the output.

```python
vague = 'Tell me about dogs.'

structured = (
    'Role: You are a veterinarian.\n'
    'Goal: Help a first-time owner choose a breed.\n'
    'Constraints: 3 breeds max, one sentence each.\n'
    'Format: a numbered list.'
)
print('VAGUE PROMPT:\n', vague)
print('\nSTRUCTURED PROMPT:\n', structured)
```

## Try it

Chain-of-thought: 'think step by step' was a prompt trick that became an architecture. Compare a bare vs reasoned prompt.

```python
problem = 'A shirt costs $40 after a 20% discount. What was the original price?'
bare = problem
cot = problem + ' Think step by step, then give the final number.'
print('Bare prompt  :', bare)
print('CoT prompt   :', cot)
print('\nThe CoT version routinely scores higher on reasoning tasks.')
```

## Takeaway

- Role + goal + constraints + format beats a vague ask.
- The frontier moved up to context engineering across whole agent runs.
