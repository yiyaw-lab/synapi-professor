# 🪟 Context Windows

How much text fits on the model's desk — and the cost of a bigger one.

```python
import numpy as np

# Self-attention compares every token to every other: cost grows as n^2.
for window in [4_000, 32_000, 128_000, 1_000_000]:
    comparisons = window ** 2
    print(f'{window:>9,} tokens -> {comparisons:,.0f} pairwise comparisons')
```

## Try it

'Lost in the middle': models attend best to the start and end. Model a crude attention budget and see where the middle gets starved.

```python
positions = np.linspace(0, 1, 11)        # 0 = start, 1 = end of context
# U-shaped attention: strong at the edges, weak in the middle
attention = 0.5 * (np.cos(2 * np.pi * positions) + 1) + 0.1
for p, a in zip(positions, attention):
    print(f'pos {p:.1f}  ' + '#' * int(a * 30))
```

## Takeaway

- Doubling the window roughly quadruples the work.
- A bigger desk does not fix attention that wanders to the edges.
