```python
import numpy as np

# Test-time compute: more reasoning samples -> higher chance one is right.
# 'Best-of-N': sample N attempts, keep the best. Accuracy climbs with N.
p_single = 0.4   # one attempt is right 40% of the time
for n in [1, 2, 4, 8, 16]:
    p_any_correct = 1 - (1 - p_single) ** n
    print(f'{n:>2} attempts -> {p_any_correct:.1%} chance at least one is correct')
```

## Try it

This is the o-series bet: accuracy scales with *thinking length*, not just model size. Change `p_single` and see how a weaker base needs more thinking to catch up.

```python
for p in [0.2, 0.4, 0.6]:
    n_needed = 1
    while 1 - (1 - p) ** n_needed < 0.9:
        n_needed += 1
    print(f'base accuracy {p:.0%} needs {n_needed:>2} attempts to reach 90%')
```

## Takeaway

- Reasoning models trade inference compute for accuracy.
- Test-time compute is a scaling axis Chinchilla never described.
