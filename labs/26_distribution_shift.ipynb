# 🎹 Distribution Shift

When live inputs drift away from what the model trained on.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q numpy
```

```python
import numpy as np

# 'Training' vocabulary the model has seen.
train_vocab = set('the cat sat on a mat dog ran log'.split())

def out_of_distribution_rate(sentence):
    words = sentence.lower().split()
    unseen = [w for w in words if w not in train_vocab]
    return len(unseen) / len(words), unseen

rate, unseen = out_of_distribution_rate('the cat sat')
print('In-distribution sentence -> OOD rate', f'{rate:.0%}', unseen)
```

## Try it

Feed it slang, a new format, or a brand-new event and watch the out-of-distribution rate spike.

```python
for s in ['the dog ran', 'rizz no cap fr fr', 'GPT-7 launched in 2029']:
    rate, unseen = out_of_distribution_rate(s)
    print(f'{s:28s} OOD {rate:>4.0%}  unseen={unseen}')
```

## Takeaway

- Models stumble when inputs drift from training data.
- Watching for that drift is core MLOps, not an afterthought.
