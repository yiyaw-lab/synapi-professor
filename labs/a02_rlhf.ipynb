# 🦮 RLHF

A reward model learns what humans applaud; the policy chases that reward.

```python
import numpy as np

# Step 1: a reward model trained on human preference pairs.
# It scores answers; here a toy version rewards being concise AND on-topic.
def reward_model(answer, on_topic):
    length_penalty = -0.02 * len(answer.split())
    return 1.0 * on_topic + length_penalty

answers = [('Paris.', 1), ('The capital is Paris.', 1), ('I love trains ' * 5, 0)]
for a, topic in answers:
    print(f'{a[:28]:30s} reward={reward_model(a, topic):+.2f}')
```

## Try it

Step 2: PPO nudges the policy toward higher reward. Simulate picking the highest-reward answer and watch the policy 'learn' what the reward model likes.

```python
best = max(answers, key=lambda x: reward_model(*x))
print('Policy converges toward:', repr(best[0]))
print('Note: it optimized the REWARD MODEL, not truth itself.')
```

## Takeaway

- RLHF = train a reward model on preferences, then RL the policy toward it.
- The policy is only ever as honest as the reward model it chases.
