# 🎓 Supervised Fine-Tuning

Turn a raw next-token predictor into an instruction-follower with examples.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
# SFT is just supervised learning on (instruction, response) pairs.
# Here: a base model continues text; SFT teaches it to STOP and answer.
dataset = [
    ('Translate to French: cat', 'chat'),
    ('Translate to French: dog', 'chien'),
    ('Translate to French: house', 'maison'),
]
# A toy 'model' = a lookup learned from examples.
model = {}
for instruction, response in dataset:
    model[instruction] = response
print('After SFT on 3 pairs, the model answers:')
print(model['Translate to French: cat'])
```

## Try it

Data quality beats quantity (the LIMA result). Add a noisy/contradictory example and watch one bad pair corrupt the behavior.

```python
dataset.append(('Translate to French: cat', 'WRONG'))  # a bad label
model = {}
for instruction, response in dataset:
    model[instruction] = response  # later examples overwrite
print("'cat' now maps to:", model['Translate to French: cat'], '<- one bad example flipped it')
```

## Takeaway

- SFT = supervised learning on instruction→response pairs.
- A handful of clean examples can beat millions of noisy ones.
