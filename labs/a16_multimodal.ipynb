# 🖼️ Multimodal Models

Project images into the same space as text — a picture becomes a kind of sentence.

```python
import numpy as np

# A VLM maps an image and its caption into a SHARED embedding space.
# Toy version: matching pairs land close; mismatches land far (CLIP idea).
np.random.seed(0)
image_cat = np.array([0.9, 0.1, 0.2])
text_cat  = np.array([0.85, 0.15, 0.25])   # caption of the same scene
text_car  = np.array([0.1, 0.9, 0.3])      # unrelated

def cos(a, b):
    return float(a @ b / (np.linalg.norm(a) * np.linalg.norm(b)))

print('image(cat) ~ text(cat):', round(cos(image_cat, text_cat), 3), '<- high, they match')
print('image(cat) ~ text(car):', round(cos(image_cat, text_car), 3), '<- low')
```

## Try it

This shared geometry is how one model reasons over pixels and words together. Add a new modality vector and check it still aligns with its caption.

```python
audio_meow = np.array([0.8, 0.2, 0.3])     # sound of the same cat
print('audio(meow) ~ text(cat):', round(cos(audio_meow, text_cat), 3))
print('Any-to-any models put image, audio, and text in one space.')
```

## Takeaway

- Multimodality = one shared embedding space across modalities.
- That alignment is what lets a model caption — or see a screen and act.
