# 🌀 Hallucinations

Confident text with no anchor in truth — and why models lack an 'I don't know'.

```python
import numpy as np

# A model always outputs a probability distribution — never silence.
def answer(confidence):
    p = np.array([confidence, (1 - confidence) / 2, (1 - confidence) / 2])
    return p

print('On a fact it knows :', np.round(answer(0.95), 2), '-> picks the top, correctly')
print('On a fact it does NOT know:', np.round(answer(0.40), 2), '-> STILL picks a top answer')
```

## Try it

Calibration: does confidence track correctness? Compare a well-calibrated model to an overconfident one.

```python
import numpy as np
stated = np.array([0.5, 0.6, 0.7, 0.8, 0.9])
actual_calibrated   = stated
actual_overconfident = stated - 0.25
print('Says 90% sure, actually right:')
print('  calibrated   ->', f'{actual_calibrated[-1]:.0%}')
print('  overconfident->', f'{actual_overconfident[-1]:.0%}  <- this gap is where hallucinations hide')
```

## Takeaway

- A model never abstains; it always emits *some* answer.
- Closing the gap between stated and real confidence is calibration.
