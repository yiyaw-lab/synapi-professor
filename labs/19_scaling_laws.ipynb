# 📈 Scaling Laws

Loss falls predictably as compute grows — a straight line in log-log.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
import numpy as np

# Toy power law: loss ~ compute^(-alpha)
compute = np.array([1, 10, 100, 1_000, 10_000], dtype=float)
loss = 5.0 * compute ** -0.1
for c, l in zip(compute, loss):
    print(f'compute x{c:>7,.0f} -> loss {l:.3f}')
```

## Try it

A power law is a straight line in log-log space. Confirm it — that straightness is what makes loss *predictable*.

```python
import numpy as np
slope = np.polyfit(np.log10(compute), np.log10(loss), 1)[0]
print('log-log slope (the scaling exponent):', round(float(slope), 3))
print('Diminishing returns: each 10x of compute buys a fixed % drop in loss.')
```

## Takeaway

- More data + compute + parameters lowers loss on a predictable curve.
