# Lab 3: Vocabulary

Every input is built from a fixed box of token pieces.

```python
import tiktoken

enc = tiktoken.get_encoding('cl100k_base')
print('Vocabulary size:', enc.n_vocab)
```

## Exercise

See how out-of-the-ordinary input fragments into many pieces.

```python
for s in ['Zbigniew', 'teh', '🦄', 'supercalifragilistic']:
    ids = enc.encode(s)
    print(repr(s), '->', len(ids), 'tokens:', [enc.decode([t]) for t in ids])
```

## Takeaway

- Rare names, typos, and emoji cost extra tokens.
- Nothing is truly 'unknown' — it is just spelled out in pieces.
