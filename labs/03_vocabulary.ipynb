```python
import tiktoken

enc = tiktoken.get_encoding('cl100k_base')
print('Vocabulary size:', enc.n_vocab)
```

## Try it

See how out-of-the-ordinary input fragments into many pieces.

```python
for s in ['Zbigniew', 'teh', '🦄', 'def f(x): return x**2']:
    ids = enc.encode(s)
    print(repr(s), '->', len(ids), 'tokens:', [enc.decode([t]) for t in ids])
```

## Takeaway

- Rare names, typos, emoji, and code cost extra tokens.
- Nothing is truly 'unknown' — it is spelled out in pieces.
