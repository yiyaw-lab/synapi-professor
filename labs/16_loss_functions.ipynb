# 📏 Loss Functions

A single number for how wrong the model just was.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q numpy
```

```python
import numpy as np

def cross_entropy(probs, correct_index):
    return -np.log(probs[correct_index])

correct = 0   # the true next token is index 0
print('confident & right :', round(cross_entropy(np.array([0.90, 0.05, 0.05]), correct), 3))
print('unsure           :', round(cross_entropy(np.array([0.34, 0.33, 0.33]), correct), 3))
print('confident & WRONG:', round(cross_entropy(np.array([0.02, 0.49, 0.49]), correct), 3))
```

## Try it

Push confidence in the wrong answer higher and watch the loss blow up.

```python
import numpy as np
for p_wrong in [0.5, 0.1, 0.01, 0.001]:
    probs = np.array([p_wrong, 1 - p_wrong])
    print(f'P(correct)={p_wrong:<6} loss={cross_entropy(probs, 0):.3f}')
```

## Takeaway

- Cross-entropy punishes confident mistakes the hardest.
- That asymmetry is what teaches a model honest uncertainty.
