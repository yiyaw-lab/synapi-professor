# 🔎 Retrieval Failures

RAG's weakest link is fetching the right passage, not writing the answer.

```python
# Keyword search misses synonyms and acronyms.
docs = {
    'd1': 'Our PTO policy grants 20 days of paid leave per year.',
    'd2': 'Submit expense reports within 30 days.',
}
def keyword_search(query):
    q = set(query.lower().split())
    return [k for k, v in docs.items() if q & set(v.lower().split())]

print("Query 'paid leave' ->", keyword_search('paid leave'))
print("Query 'vacation days' ->", keyword_search('vacation days'), '<- misses! synonym gap')
```

## Try it

Fix it with a tiny synonym map (a stand-in for embeddings / hybrid search).

```python
SYNONYMS = {'vacation': 'paid leave', 'holiday': 'paid leave', 'pto': 'paid leave'}

def expand(query):
    return ' '.join(SYNONYMS.get(w, w) for w in query.lower().split())

fixed = expand('vacation days')
print('Expanded query:', repr(fixed))
print("Now retrieves ->", keyword_search(fixed))
```

## Takeaway

- Most 'the AI was wrong' bugs are really retrieval bugs.
- Synonyms, hybrid search, and rerankers close the gap.
