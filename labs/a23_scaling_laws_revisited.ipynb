# 📐 Scaling Laws Revisited

Chinchilla: most big models were undertrained — too many params, too little data.

```python
import numpy as np

# Toy loss from parameters N and data D, under a fixed compute budget C ~ N*D.
def loss(N, D):
    return 1.0 + 0.5 * N ** -0.34 + 0.5 * D ** -0.28

C = 1e6
best = min(
    ((N, C / N) for N in np.logspace(1, 5, 50)),
    key=lambda nd: loss(*nd),
)
print(f'Compute-optimal split: ~{best[0]:.0f} params, ~{best[1]:.0f} data')
print(f'Minimum loss: {loss(*best):.4f}')
```

## Try it

Inference economics flips it: if you serve a model billions of times, OVERtrain a smaller one (the Llama 3 bet). Compare a balanced vs. a small-but-overtrained model.

```python
balanced = loss(best[0], best[1])
small_overtrained = loss(best[0] / 4, best[1] * 4)
print(f'compute-optimal loss   : {balanced:.4f}')
print(f'small + overtrained    : {small_overtrained:.4f}  (slightly worse loss,')
print('                          but far cheaper to SERVE billions of times)')
```

## Takeaway

- Chinchilla rebalanced params vs. data toward more data.
- The optimum depends on whether you pay to train or to serve.
