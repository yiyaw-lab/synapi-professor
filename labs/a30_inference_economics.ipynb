# 💸 Inference Economics

Throughput, latency, and cost — where serving LLMs is won or lost.

```python
# Cost of one feature: tokens/request x requests/day x price/token.
tokens_per_request = 2_000
requests_per_day = 50_000
price_per_1m_tokens = 3.00

daily = tokens_per_request * requests_per_day / 1_000_000 * price_per_1m_tokens
print(f'Daily token cost: ${daily:,.2f}')
print(f'Monthly        : ${daily * 30:,.2f}')
```

## Try it

Batching and a smaller model are the big levers. Show how each changes the bill — and circle the one that makes the feature ship-able.

```python
# A cheaper model at 1/5 the price, or batching that lifts effective throughput.
for label, price in [('frontier $3/1M', 3.00), ('small $0.60/1M', 0.60)]:
    cost = tokens_per_request * requests_per_day / 1_000_000 * price * 30
    print(f'{label:18s} -> ${cost:>10,.2f}/month')
print('Continuous batching + paged attention lift tokens-per-dollar without changing the model.')
```

## Takeaway

- Inference, not training, is where most AI compute and money go.
- 'Tokens per dollar' decides which products are even viable.
