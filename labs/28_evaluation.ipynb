```python
# A question whose answer can be reached by copying the prompt — a weak eval.
def grade(answer, gold):
    return gold.lower() in answer.lower()

prompt = 'The capital of France is Paris. What is the capital of France?'
parrot = 'The capital of France is Paris.'
print('Parrot passes the weak eval:', grade(parrot, 'paris'), '<- but proves nothing')
```

## Try it

Write a question that *cannot* be answered by paraphrasing the prompt, then grade two answers.

```python
question = 'If a train leaves at 2pm going 60mph, where is it at 4pm?'
gold = '120 miles'
answers = {'reasoned': 'It has gone 120 miles.', 'fluent-but-wrong': 'Somewhere far away.'}
for who, a in answers.items():
    print(f'{who:18s} -> {"PASS" if grade(a, gold) else "FAIL"}')
```

## Takeaway

- Saturating, leaky benchmarks pushed eval toward harder, live tests.
- A good eval separates understanding from fluent-sounding nonsense.
