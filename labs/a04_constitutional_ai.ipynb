# 📜 Constitutional AI

The model critiques and revises its own answer against written principles.

```python
CONSTITUTION = [
    'Do not give instructions that enable serious harm.',
    'Be honest about uncertainty.',
]

def critique(draft):
    issues = []
    if any(w in draft.lower() for w in ['weapon', 'explosive']):
        issues.append(CONSTITUTION[0])
    return issues

draft = 'Sure, here is how to build a weapon at home.'
print('Critique found:', critique(draft))
```

## Try it

Now run the revise step: rewrite the draft until it passes the critique. That critique→revise loop is RLAIF in miniature.

```python
def revise(draft):
    if critique(draft):
        return 'I can\'t help with that, but here is a safe, legal alternative.'
    return draft

revised = revise(draft)
print('Revised:', revised)
print('Passes now:', critique(revised) == [])
```

## Takeaway

- The model supervises itself against a written constitution.
- AI feedback can stand in for human labels — the seed of scalable oversight.
