# 🎟️ Positional Encoding

Transformers see tokens in parallel — position is what restores order.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
import numpy as np

def positional_encoding(seq_len, d_model):
    pos = np.arange(seq_len)[:, None]
    i = np.arange(d_model)[None, :]
    angle = pos / np.power(10000, (2 * (i // 2)) / d_model)
    enc = np.where(i % 2 == 0, np.sin(angle), np.cos(angle))
    return enc

pe = positional_encoding(seq_len=6, d_model=8)
print('Position 0:', np.round(pe[0], 2))
print('Position 1:', np.round(pe[1], 2))
print('Each position gets a unique fingerprint of sines and cosines.')
```

## Try it

'dog bites man' vs 'man bites dog' — same tokens, different positions. Confirm the position vectors differ.

```python
import numpy as np
same_token = np.array([0.2, -0.5, 0.1, 0.7, -0.3, 0.4, 0.0, -0.2])
at_pos0 = same_token + pe[0]
at_pos2 = same_token + pe[2]
print('Same word at position 0 vs 2 differs:', not np.allclose(at_pos0, at_pos2))
print('Distance between them:', round(float(np.linalg.norm(at_pos0 - at_pos2)), 3))
```

## Takeaway

- Position is injected, not assumed.
- Identical tokens at different places become different vectors — that is meaning.
