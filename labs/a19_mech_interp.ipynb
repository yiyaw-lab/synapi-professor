```python
import numpy as np

# 'Induction heads' do in-context pattern completion: ...AB...A -> B.
# Toy version: find where a token last appeared and predict what followed.
seq = list('ABCAB')
def induction_predict(seq):
    last = seq[-1]
    for i in range(len(seq) - 2, -1, -1):
        if seq[i] == last:
            return seq[i + 1]   # 'attend back to A, copy what followed'
    return '?'

print('sequence:', ''.join(seq))
print('induction head predicts next:', induction_predict(seq), '(saw A->B before)')
```

## Try it

This tiny circuit is a real, findable algorithm inside transformers. Feed it a new pattern and confirm it generalizes.

```python
for s in ['XYXY'[:3] + 'X', 'helloh', '1212']:
    print(f'{s:8s} -> {induction_predict(list(s))}')
print('Real induction heads were *found* in trained models — not designed in.')
```
