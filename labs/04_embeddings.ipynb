# Lab 4: Embeddings

Tokens become vectors that capture meaning.

_Requires: `pip install sentence-transformers`_

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
vecs = model.encode(['king', 'queen', 'banana'])
print('Vector dimension:', vecs.shape[1])
print('First 5 dims of king:', vecs[0][:5])
```

## Exercise

Embed words of your choice and inspect the shapes.

```python
words = ['paris', 'france', 'tokyo', 'sushi']
vecs = model.encode(words)
for w, v in zip(words, vecs):
    print(f'{w:8s} -> dim {len(v)}, norm {float((v**2).sum()**0.5):.3f}')
```

## Takeaway

- Each token/word maps to a list of numbers.
- Those numbers encode learned patterns of meaning.
