```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
vecs = model.encode(['king', 'queen', 'banana'])
print('Vector dimension:', vecs.shape[1])
print('First 5 dims of king:', vecs[0][:5])
```

## Try it

Build a tiny semantic search: which sentence is closest to your query?

```python
from sentence_transformers import util
corpus = ['The cat naps in the sun.', 'Quantum computers use qubits.',
          'I love a strong espresso.', 'The dog chased the ball.']
q = 'a sleepy kitten'
scores = util.cos_sim(model.encode(q), model.encode(corpus))[0]
best = int(scores.argmax())
print('Query:', q)
print('Closest:', corpus[best], f'({float(scores[best]):.3f})')
```

## Takeaway

- Each word/sentence maps to a list of numbers.
- Those numbers power every RAG and search product.
