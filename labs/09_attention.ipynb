# 👁️ Attention

Attention decides which tokens matter for the current one.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q numpy
```

```python
import numpy as np

def softmax(x):
    e = np.exp(x - x.max())
    return e / e.sum()

# toy 'scores' of how much a query token relates to 4 context tokens
scores = np.array([2.0, 0.1, 1.0, -1.0])
weights = softmax(scores)
print('Attention weights:', np.round(weights, 3))
print('They sum to:', round(float(weights.sum()), 3))
```

## Try it

Change the scores and watch attention concentrate or spread.

```python
tokens = ['the', 'animal', 'street', 'tired']
for tok, w in zip(tokens, weights):
    print(f'{tok:8s} {w:.3f} ' + '#' * int(w * 40))
```

## Takeaway

- Softmax turns raw scores into weights that sum to 1.
- Attention is a weighted average over context.
