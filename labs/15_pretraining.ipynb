# 📚 Pretraining

Learning language by predicting the next token, billions of times.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
from collections import Counter, defaultdict

# A tiny 'pretraining' corpus. The model: a next-word frequency table.
corpus = ('the cat sat on the mat the cat ran to the dog '
          'the dog sat on the log the cat sat again').split()
model = defaultdict(Counter)
for a, b in zip(corpus, corpus[1:]):
    model[a][b] += 1
print("After 'the', the model expects:", dict(model['the']))
```

## Try it

Generate text by always taking the most likely next word — knowledge that emerged purely from counting.

```python
word, out = 'the', ['the']
for _ in range(8):
    if not model[word]:
        break
    word = model[word].most_common(1)[0][0]
    out.append(word)
print(' '.join(out))
```

## Takeaway

- Pretraining is next-token prediction at planetary scale.
- 'Knowledge' is just compressed statistics of what tends to follow what.
