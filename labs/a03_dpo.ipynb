# ⚖️ Direct Preference Optimization

Skip the reward model and the RL loop — one loss straight from 'A beat B'.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# DPO loss: -log sigmoid( beta * (chosen_logratio - rejected_logratio) ).
# logratio = log p_policy(y) - log p_reference(y).
def dpo_loss(chosen_logratio, rejected_logratio, beta=0.1):
    return -np.log(sigmoid(beta * (chosen_logratio - rejected_logratio)))

# When the policy already prefers the chosen answer, loss is low.
print('chosen >> rejected :', round(float(dpo_loss(2.0, -2.0)), 3))
print('chosen ~= rejected :', round(float(dpo_loss(0.0, 0.0)), 3))
print('chosen << rejected :', round(float(dpo_loss(-2.0, 2.0)), 3), '<- biggest gradient')
```

## Try it

Turn the temperature `beta` up and down. It controls how hard DPO pushes the chosen answer above the rejected one.

```python
for beta in [0.05, 0.1, 0.5, 1.0]:
    loss = float(dpo_loss(-1.0, 1.0, beta=beta))
    print(f'beta={beta:<4}  loss={loss:.3f}')
```

## Takeaway

- DPO turns 'preferred > rejected' directly into one classification-style loss.
- No reward model, no PPO — which is why it took over open-model alignment.
