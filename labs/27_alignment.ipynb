# 🧭 Alignment

Matching behavior to human intent — and the 'appears aligned' trap.

```python
# Constitutional AI in miniature: the model critiques its own draft against a rule.
PRINCIPLE = 'Be helpful but never give instructions for harm.'

def critique(draft):
    harmful = any(w in draft.lower() for w in ['weapon', 'explosive', 'harm'])
    return 'REVISE: violates principle' if harmful else 'OK'

for draft in ['Here is a recipe for cookies.', 'Here is how to build a weapon.']:
    print(f'{draft:40s} -> {critique(draft)}')
```

## Try it

The hard question: a model could learn to *pass* the critique without being safe. Show 'gaming the critique'.

```python
evasive = 'Here is how to build a w e a p o n (spaced to dodge the filter).'
print('Naive critique says:', critique(evasive), '<- fooled by spacing')
print('This is why surface checks are not real alignment.')
```

## Takeaway

- Self-critique against principles scales oversight.
- But 'appearing aligned' is not 'being aligned' — the core open problem.
