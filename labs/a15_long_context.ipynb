# 📜 Long-Context Methods

Stretch a short-trained model to long context by rescaling positions.

```python
import numpy as np

# RoPE encodes position as rotation angles. Position interpolation divides
# the position by a scale factor, squeezing long positions into the
# trained range so the model 'recognizes' them.
def rope_angle(pos, dim_idx, d=64):
    return pos / (10000 ** (2 * dim_idx / d))

trained_len, target_len = 2048, 8192
scale = target_len / trained_len
pos = 6000  # beyond training length
print('raw angle at pos 6000 :', round(rope_angle(pos, 1), 3), '(out of trained range)')
print('interpolated angle    :', round(rope_angle(pos / scale, 1), 3), '(back in range)')
```

## Try it

Interpolation lets an 8k-trained model handle far longer. But 'lost in the middle' persists — model a U-shaped attention budget over positions.

```python
positions = np.linspace(0, 1, 11)
attention = 0.5 * (np.cos(2 * np.pi * positions) + 1) + 0.1
for p, a in zip(positions, attention):
    print(f'pos {p:.1f}  ' + '#' * int(a * 30))
print('A bigger window does not fix attention that wanders to the edges.')
```

## Takeaway

- Rescaling positions extends context without full retraining.
- Capacity to read long != actually using the middle.
