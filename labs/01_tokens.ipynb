# Lab 1: Tokens

LLMs read tokens, not words. See what the pieces actually look like.

```python
import tiktoken

enc = tiktoken.get_encoding('cl100k_base')
text = 'I love learning.'
ids = enc.encode(text)
print('Text:', text)
print('Token IDs:', ids)
print('Pieces:', [enc.decode([t]) for t in ids])
```

## Exercise

Guess how each phrase splits, then run it.

```python
for phrase in ['cat', 'concatenate', 'antidisestablishmentarianism']:
    ids = enc.encode(phrase)
    print(phrase, '->', [enc.decode([t]) for t in ids])
```

## Takeaway

- A 'word' is often several tokens.
- Token count drives cost and context limits.
