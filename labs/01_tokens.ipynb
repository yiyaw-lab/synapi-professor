# 🔤 Tokens

LLMs read tokens, not words. Let's see the pieces — and what they cost.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q tiktoken
```

```python
import tiktoken

enc = tiktoken.get_encoding('cl100k_base')
text = 'I love learning.'
ids = enc.encode(text)
print('Text:', text)
print('Token IDs:', ids)
print('Pieces:', [enc.decode([t]) for t in ids])
```

## Try it

Guess how each phrase splits, then run it. Then peek at the cost.

```python
for phrase in ['cat', 'concatenate', 'antidisestablishmentarianism']:
    ids = enc.encode(phrase)
    print(phrase, '->', [enc.decode([t]) for t in ids])

# rough cost: GPT-4o input ~ $2.50 / 1M tokens
doc = 'token economics ' * 200
n = len(enc.encode(doc))
print(f'\n{n} tokens  ~  $' + format(n / 1_000_000 * 2.50, '.6f') + ' to send once')
```

## Takeaway

- A 'word' is often several tokens.
- Token count *is* the bill and the context limit.
