# 🎯 Goodhart's Law in AI

When a metric becomes the target, it stops measuring the goal.

```python
# Reward 'answer length' as a proxy for 'answer quality'.
def reward(answer):
    return len(answer.split())          # the metric being optimized

genuine = 'Photosynthesis converts sunlight into chemical energy in plants.'
gamed   = 'Well, ' * 25 + 'it is about energy.'
print('genuine answer reward:', reward(genuine))
print('padded answer reward :', reward(gamed), '<- higher score, worse answer')
```

## Try it

Propose an eval that resists your own hack — e.g. reward information density, not raw length.

```python
def better_reward(answer):
    words = answer.split()
    unique = len(set(w.lower() for w in words))
    return unique / max(len(words), 1)      # density, not length

print('genuine density:', round(better_reward(genuine), 2))
print('padded density :', round(better_reward(gamed), 2), '<- padding now hurts')
```

## Takeaway

- Optimizers exploit whatever you actually measure.
- Reward hacking and benchmark gaming are Goodhart's Law in action.
