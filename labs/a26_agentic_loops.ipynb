# 🔁 Agentic Loops

Reason, act, observe, repeat — until the goal is met. And watch errors compound.

```python
# A minimal ReAct loop: Thought -> Action -> Observation, repeated.
def fake_tool(query):
    facts = {'population of France': '68 million', 'capital of France': 'Paris'}
    return facts.get(query, 'unknown')

goal = 'population of France'
for step in range(1, 4):
    thought = f'I should look up the {goal}.'
    observation = fake_tool(goal)
    print(f'Step {step}: THOUGHT {thought}\n        OBSERVATION {observation}')
    if observation != 'unknown':
        print('        DONE.')
        break
```

## Try it

The hard part is error compounding over long horizons. Inject one wrong observation and watch every later step inherit it.

```python
def buggy_tool(query):
    return 'population of France' in query and '999 billion' or 'unknown'

obs = buggy_tool('population of France')
print('Step 1 observation (wrong):', obs)
print('Step 2 reasons FROM that wrong number...')
print('Conclusion:', f'France has {obs} people. <- one bad step poisons the whole chain')
```

## Takeaway

- Agents loop reason→act→observe instead of answering once.
- Over long horizons, evaluation (not generation) is the bottleneck.
