# 🌌 Semantic Space

Meanings live as positions; related ideas orbit close.

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer('all-MiniLM-L6-v2')
anchor = 'school'
words = ['teacher', 'student', 'exam', 'classroom', 'banana', 'rocket']
scores = util.cos_sim(model.encode(anchor), model.encode(words))[0]
for w, s in sorted(zip(words, scores), key=lambda x: -x[1]):
    print(f'{w:10s} {float(s):.3f}')
```

## Try it

Change the anchor to your own word and re-run. Where does 'banana' land now?

```python
anchor = 'ocean'  # <- change me
words = ['wave', 'salt', 'whale', 'desert', 'guitar']
for w, s in sorted(zip(words, util.cos_sim(model.encode(anchor), model.encode(words))[0]), key=lambda x: -x[1]):
    print(f'{w:10s} {float(s):.3f}')
```

## Takeaway

- Distance tracks relatedness.
- 'banana' and 'rocket' fall far from 'school'.
