# Lab 5: Semantic Space

Meanings live as positions; related ideas sit closer.

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer('all-MiniLM-L6-v2')
anchor = 'school'
words = ['teacher', 'student', 'exam', 'classroom', 'banana', 'rocket']
av = model.encode(anchor)
wv = model.encode(words)
scores = util.cos_sim(av, wv)[0]
for w, s in sorted(zip(words, scores), key=lambda x: -x[1]):
    print(f'{w:10s} {float(s):.3f}')
```
