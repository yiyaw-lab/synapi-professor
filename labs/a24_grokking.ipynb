# 💡 Grokking

Long after memorizing, a model suddenly *generalizes* — the rule clicks.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q numpy
```

```python
import numpy as np

# Simulate grokking: train accuracy saturates early; test accuracy lags,
# then jumps long after 'convergence'.
epochs = np.arange(0, 1000, 50)
train_acc = np.minimum(1.0, epochs / 100)              # memorizes fast
test_acc = 1 / (1 + np.exp(-(epochs - 700) / 60))      # generalizes LATE
for e, tr, te in list(zip(epochs, train_acc, test_acc))[::3]:
    print(f'epoch {e:>4}: train {tr:.2f}  test {te:.2f}')
```

## Try it

Find the moment the model 'groks' — where test accuracy finally crosses 0.9, far past where training looked done.

```python
grok_epoch = next(e for e, te in zip(epochs, test_acc) if te > 0.9)
print(f'Train looked solved by epoch ~100, but grokking happens at epoch {grok_epoch}.')
print('"Done training" is a blurrier line than the loss curve suggests.')
```

## Takeaway

- Generalization can arrive long after memorization.
- A model can look finished while one step from truly understanding.
