# Lab 2: Tokenization

Why words get chopped into sub-word pieces.

```python
import tiktoken

enc = tiktoken.get_encoding('cl100k_base')
for word in ['believable', 'unbelievable', 'unbelievably']:
    ids = enc.encode(word)
    print(f'{word:16s} {len(ids)} tokens -> {[enc.decode([t]) for t in ids]}')
```

## Exercise

Find a long word that is a single token, and a short one that is several.

```python
tests = ['the', ' the', 'hello', 'GPT', 'asdfqwer', '日本語']
for t in tests:
    ids = enc.encode(t)
    print(repr(t), '->', len(ids), 'tokens')
```

## Takeaway

- Common strings get one token; rare ones get many.
- Leading spaces and case change the split.
