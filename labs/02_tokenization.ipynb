# 🔪 Tokenization

Why words get chopped into sub-word pieces — and who pays for it.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q tiktoken
```

```python
import tiktoken

enc = tiktoken.get_encoding('cl100k_base')
for word in ['believable', 'unbelievable', 'unbelievably']:
    ids = enc.encode(word)
    print(f'{word:16s} {len(ids)} tokens -> {[enc.decode([t]) for t in ids]}')
```

## Try it

The SAME sentence costs more tokens in some languages. Watch the gap.

```python
pairs = {
    'English': 'The cat sleeps on the warm windowsill.',
    'Hindi':   'बिल्ली गर्म खिड़की पर सोती है।',
    'Thai':    'แมวนอนอยู่บนขอบหน้าต่างที่อบอุ่น',
}
for lang, s in pairs.items():
    print(f'{lang:8s} {len(enc.encode(s)):3d} tokens')
```

## Takeaway

- Common strings get one token; rare ones get many.
- Tokenizer choices decide whose language is cheap.
