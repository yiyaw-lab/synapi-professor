# ⛰️ Gradient Descent

Walking downhill in the fog to find the bottom of the loss.

> ▶︎ In Colab: **Runtime → Run all** — this notebook runs top to bottom with no setup.

```python
%pip install -q numpy
```

```python
# Minimize f(x) = x^2, whose slope is 2x. Start at x = 4.
x, lr = 4.0, 0.1
for step in range(6):
    grad = 2 * x
    x = x - lr * grad
    print(f'step {step}: x={x:.4f}  f(x)={x**2:.4f}')
```

## Try it

Sabotage it: crank the learning rate too high and watch the steps explode instead of settle.

```python
x, lr = 4.0, 1.1     # too big
for step in range(6):
    x = x - lr * (2 * x)
    print(f'step {step}: x={x:.2f}  f(x)={x**2:.2f}')
```

## Takeaway

- Each step nudges parameters down the slope of the loss.
- Learning rate is the whole game: too small crawls, too big diverges.
