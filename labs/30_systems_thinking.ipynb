```python
# Map a real AI product as a system and find its weakest link.
system = {
    'model':    {'role': 'generate answers',  'failure': 'hallucination'},
    'data':     {'role': 'ground answers',    'failure': 'stale / wrong docs'},
    'retrieval':{'role': 'fetch context',     'failure': 'misses the passage'},
    'user':     {'role': 'asks questions',    'failure': 'ambiguous prompt'},
    'feedback': {'role': 'learn from use',    'failure': 'no signal collected'},
}
for part, info in system.items():
    print(f'{part:10s} -> {info["role"]:20s} | fails by: {info["failure"]}')
```

## Try it

Circle the weakest link and propose one concrete fix — that is the systems-thinking move.

```python
weakest = 'retrieval'
fix = 'add a reranker + hybrid search; log retrieval hit-rate as a metric'
print(f'Weakest link: {weakest}')
print(f'Failure     : {system[weakest]["failure"]}')
print(f'Proposed fix: {fix}')
```

## Takeaway

- An LLM is an engine; it needs steering, fuel, and brakes.
- The hard 2025 problems are systems problems: orchestration, cost, trust.
