# ♻️ Synthetic Data

Models generate their own training data — powerful, and quietly risky.

```python
import numpy as np

np.random.seed(0)
# Real data has a true spread. Train a model, sample from it, train again...
real = np.random.normal(0, 1, 10_000)
print(f'real data      : mean={real.mean():.2f} std={real.std():.2f}')

# Each generation samples from the previous (slightly narrowed) one.
data = real
for gen in range(1, 4):
    data = np.random.normal(data.mean(), data.std() * 0.85, 10_000)
    print(f'gen {gen} synthetic: mean={data.mean():.2f} std={data.std():.2f}')
```

## Try it

Watch the variance shrink each generation — that's model collapse: the tails (rare knowledge) vanish first. Run more generations.

```python
data = real
for gen in range(1, 8):
    data = np.random.normal(data.mean(), data.std() * 0.85, 10_000)
print(f'after 7 generations: std={data.std():.3f} (started at 1.00 — the diversity is gone)')
```

## Takeaway

- Synthetic data scales past the human-text 'data wall'.
- Trained on carelessly, models collapse toward the bland middle.
