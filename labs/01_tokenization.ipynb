# Lab 1: Tokenization

This notebook demonstrates how tokenization works and why token boundaries matter for LLMs.

```python
import tiktoken

encoder = tiktoken.get_encoding('cl100k_base')
text = 'I love learning.'
tokens = encoder.encode(text)
print('Text:', text)
print('Token IDs:', tokens)
print('Decoded tokens:', [encoder.decode([t]) for t in tokens])
```

## Exercise

Try changing the sentence and observe how the tokenization differs. What happens with punctuation, capitalization, or rare words?

```python
for phrase in [
    'unbelievable',
    'I love language models.',
    'ChatGPT is amazing!',
]:
    token_ids = encoder.encode(phrase)
    decoded = [encoder.decode([t]) for t in token_ids]
    print('---')
    print('Phrase:', phrase)
    print('Token IDs:', token_ids)
    print('Tokens:', decoded)
```

## Expected takeaway

- Tokenizers can split a single word into multiple pieces.
- Rare words and punctuation change token boundaries.
- Token count matters for pricing, context windows, and model behavior.
