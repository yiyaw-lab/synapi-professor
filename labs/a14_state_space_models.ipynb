# 🌀 State-Space Models

Carry a compressed state forward in linear time — no all-pairs attention.

```python
import numpy as np

# An SSM scans the sequence, updating a fixed-size state: h = a*h + b*x.
def ssm_scan(xs, a=0.9, b=1.0):
    h, outputs = 0.0, []
    for x in xs:
        h = a * h + b * x        # O(1) per step, O(n) total
        outputs.append(h)
    return outputs

xs = [1, 0, 0, 0, 0, 0]          # an impulse
print('state over time:', [round(o, 3) for o in ssm_scan(xs)])
print('The past fades into a single carried state — no N x N matrix.')
```

## Try it

The decay `a` is the model's memory. Closer to 1 = remembers longer. Compare linear SSM cost to attention's quadratic cost.

```python
for n in [1_000, 10_000, 100_000]:
    print(f'seq {n:>7,}: attention ~ {n**2:>15,} ops | SSM ~ {n:>10,} ops')
```

## Takeaway

- SSMs scale linearly by compressing the past into a state.
- Attention re-reads everything; SSMs summarize — and hybrids may win.
